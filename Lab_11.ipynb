```python
# Data Wrangling
# Combining data using merge and join
```

```python
student =  pd.DataFrame({
    'Student_id' : [101,102,103,104],
    'Name' : ['Aman','Priya','Rahul','Shagun'],
    'Dept' : ["CSE","CSE","DS","ECE"]
})

marks =  pd.DataFrame({
    'Student_id' : [101,102,104,105],
    'Marks' : [82,72,91,86]
})
```

```python
print("Student table:")
print(student)
print("\nMarks table:")
print(marks)
```

```python
inner = pd.merge(student,marks,on='Student_id',how='inner')
print("\nInner join:")
print(inner)
```

```python
right = pd.merge(student,marks,on='Student_id',how='right')
print("\nRight join:")
print(right)
```

```python
outer = pd.merge(student,marks,on='Student_id',how='outer')
print("\nOuter join:")
print(outer)
```

```python
# Column names differ
```

```python
student_info = student.rename(columns={'Student_id':'ID'})
print("\nStudent Information:")
print(student_info)

combined = pd.merge(student_info,marks,left_on='ID',right_on='Student_id',how='inner')
print("\nCombined Data:")
print(combined)
```

```python
# Join using an index
```

```python
students_indexed = student.set_index('Student_id')
marks_indexed = marks.set_index('Student_id')

joined = students_indexed.join(marks_indexed,how='left')
print("\nStudents (Indexed):")
print(students_indexed)
print("\nMarks (Indexed):")
print(marks_indexed)
print("\nJoined:")
print(joined)
```

```python
import pandas as pd
import numpy as np

raw_students = pd.DataFrame({
    "Roll_No": [1, 2, 3, 3, 4, 5],
    "Name": [" aman ", "PRIYA", "Rahul", "Rahul", "Neha", "Isha"],
    "Department": ["cse", "IT", "CSE", "CSE", "ece", "IT"],
    "Marks": ["85", "90", "78", "78", "missing", "67"],
    "Attendance": [92, 95, 80, 80, 88, np.nan]
})
print(raw_students)
```

```python
# 1. Inspect shape, types, missing values, duplicates
print("Shape:", raw_students.shape)
print(raw_students.dtypes)
print("Missing values:\n", raw_students.isna().sum())
print("Duplicate rows:", raw_students.duplicated().sum())
print("Duplicate Roll_No:", raw_students["Roll_No"].duplicated().sum())

df = raw_students.copy()
```

```python
# 2. Standardize names and department labels
df["Name"] = df["Name"].str.strip().str.title()
df["Department"] = df["Department"].str.strip().str.upper()
```

```python
# 3. Convert Marks to numeric; 'missing' becomes NaN
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
```

```python
# 4. Remove duplicates using Roll_No as the key
df = df.drop_duplicates(subset="Roll_No", keep="first").reset_index(drop=True)
```

```python
# 5. Handle missing values
# Attendance: fill with the department median (fallback: overall median)
df["Attendance"] = df["Attendance"].fillna(
    df.groupby("Department")["Attendance"].transform("median")
).fillna(df["Attendance"].median())

# Marks: keep as NaN (do not invent a score); flag them instead
df["Marks_Missing"] = df["Marks"].isna()
```

```python
# 6. Pass column (threshold = 40)
df["Pass"] = np.where(
    df["Marks"].isna(), "Not Evaluated",
    np.where(df["Marks"] >= 40, "Pass", "Fail")
)

print(df)
```

```python
# 7. Average marks and attendance by department
dept_summary = df.groupby("Department")[["Marks", "Attendance"]].mean().round(2)
print(dept_summary)
```

```python
# 8. Save as CSV and Pickle
df.to_csv("cleaned_students.csv", index=False)
df.to_pickle("cleaned_students.pkl")
```

```python
import pandas as pd

# DataFrame 1
df1 = pd.DataFrame({
    "Movie_ID": [1, 2, 3, 4, 5, 6],
    "Movie_Name": ["Inception", "Interstellar", "Titanic", "Avatar", "Joker", "Tenet"],
    "Genre": ["Sci-Fi", "Sci-Fi", "Romance", "Fantasy", "Drama", "Action"]
})

# DataFrame 2
df2 = pd.DataFrame({
    "Movie_ID": [1, 2, 3, 4, 5, 7],
    "Rating": [4.8, 4.6, 4.1, 3.9, 4.4, 4.2],
    "Number_of_Reviews": [2500, 3100, 1800, 2200, 1500, 900]
})

# Merge on Movie_ID
merged = pd.merge(df1, df2, on="Movie_ID")
print(merged)

# Movies with rating > 4
print(merged[merged["Rating"] > 4])
```

```python
outer = pd.merge(df1, df2, on="Movie_ID", how="outer", indicator=True)
print(outer)

# Rows that exist in only one table
unmatched = outer[outer["_merge"] != "both"]
print(unmatched[["Movie_ID", "_merge"]])
print("Non-matching IDs:", unmatched["Movie_ID"].tolist())
```

```python
# Works: both Movie_ID columns are unique
pd.merge(df1, df2, on="Movie_ID", validate="one_to_one")
```

```python
# merge(): joins on columns (or index), flexible keys
m = pd.merge(df1, df2, on="Movie_ID")

# join(): joins on the INDEX by default
j = df1.set_index("Movie_ID").join(df2.set_index("Movie_ID"))
print(j)
```
